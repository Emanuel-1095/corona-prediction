```python
import pandas as pd
import re


def binary_proteins_csv(filename):
    """
    Transforma los valores de las proteinas en valores binario
    si el valor de una proteina es 0, entonces le coloca 0, en otro caso le coloca el valor 1
    lo utilizamos para poder utilizar la distancia jaccard para la proyección UMAP y para el K-means
    """
    # 1. Cargar el dataset
    # Asegúrate de que el nombre del archivo sea el correcto
    df = pd.read_csv(filename + '.csv')

    # 2. Identificar columnas usando Expresiones Regulares (Regex)
    # El patrón r'^[POQ]' significa: "Cualquier texto que empiece (^) con P, O, o Q"
    # Esto cumple tu requisito para detectar las proteínas dinámicamente.
    columnas_proteinas = [col for col in df.columns if re.match(r'^[POQ]', col)]
    
    # Las columnas de metadatos son todas las que NO están en la lista de proteínas
    columnas_metadatos = [col for col in df.columns if col not in columnas_proteinas]
    
    print(f"Procesando: {len(columnas_metadatos)} columnas de metadatos y {len(columnas_proteinas)} de proteínas.")

    # 3. Separar los datos
    df_metadatos = df[columnas_metadatos]
    df_proteinas = df[columnas_proteinas]

    # 4. Aplicar la transformación booleana a las proteínas
    # (Valores > 0 se convierten en 1, el resto en 0)
    df_proteinas_binario = (df_proteinas > 0).astype(int)

    # 5. Unir nuevamente
    df_resultado = pd.concat([df_metadatos, df_proteinas_binario], axis=1)

    df_resultado.to_csv(filename+'_binary.csv', index=False)
    print(f"Archivo guardado exitosamente como: {filename+'_binary.csv'}")
    
```

```python
binary_proteins_csv('dataset_total_filtrado_frecuentes')
```
