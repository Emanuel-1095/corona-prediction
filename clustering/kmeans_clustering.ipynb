```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

"""
1. Carga los datos binarios de proteínas.
2. Ejecuta un ciclo iterativo del algoritmo K-Means variando la cantidad de clusters (k) entre 240 y 400.
3. Evalúa la calidad de cada modelo usando el 'Silhouette Score' con métrica Jaccard (ideal para datos binarios).
4. Grafica los puntajes para identificar visualmente el número óptimo de clusters (donde la silueta es mayor).
"""

# 1. Cargar el dataset binario
df = pd.read_csv('individual_proteins_dataset_binary.csv')

# 2. Seleccionar solo las columnas de proteínas (datos binarios)
# Ajustamos el índice según tu dataset (las proteínas empiezan después de 'modification_type')
start_col = df.columns.get_loc("modification_type") + 1
X = df.iloc[:, start_col:]

# 3. Configuración para probar diferentes valores de k (número de clusters)
range_n_clusters = range(2, 250)
silhouette_avg_scores = []

print("Calculando clusters y siluetas...")

for k in range_n_clusters:
    # Inicializar KMeans
    # n_init=10 ejecuta el algoritmo 10 veces con diferentes centroides iniciales
    kmeans = KMeans(n_clusters=k, random_state=42)
    
    # Predecir los clusters
    cluster_labels = kmeans.fit_predict(X)
    
    # 4. Calcular el coeficiente de silueta usando la métrica JACCARD
    # Esta función compara qué tan similar es un objeto a su propio cluster (cohesión)
    # versus otros clusters (separación), usando la distancia de Jaccard.
    silhouette_avg = silhouette_score(X, cluster_labels, metric='jaccard')
    silhouette_avg_scores.append(silhouette_avg)
    
    print(f"Para n_clusters = {k}, el puntaje de silueta (Jaccard) es: {silhouette_avg:.4f}")

# 5. Graficar los resultados para encontrar el k óptimo
plt.figure(figsize=(10, 6))
plt.plot(range_n_clusters, silhouette_avg_scores, marker='o', linestyle='-', color='b')
plt.title('Puntaje de Silueta (Jaccard) vs. Número de Clusters')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Coeficiente de Silueta (Jaccard)')
plt.grid(True)
plt.show()
```
