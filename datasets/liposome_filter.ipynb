```python
import pandas as pd
```

```python
def filtrar_liposomas(archivo_entrada, archivo_salida, atributo_cat, categoria):
    """
    Filtra un dataset para mantener solo las filas donde 'np_type' es 'liposome',
    elimina la columna 'np_type' para que no aparezca en el resultado final, 
    y guarda el resultado en un nuevo archivo CSV.
    """
    # 1. Leer el dataset original
    df = pd.read_csv(archivo_entrada)
    
    # 2. Filtrar por el atributo 'np_type'
    # Usamos .copy() para evitar advertencias (SettingWithCopyWarning) al modificar el dataframe luego
    df_filtrado = df[df[atributo_cat] == categoria].copy()
    
    # 3. Mostrar cuántos registros se encontraron
    print(f"Se encontraron {len(df_filtrado)} registros de liposomas.")
    
    # 4. Eliminar la columna 'np_type' del dataframe filtrado
    df_filtrado = df_filtrado.drop(columns=[atributo_cat])
    
    # 5. Guardar el DataFrame filtrado (ya sin la columna) en un nuevo archivo CSV
    df_filtrado.to_csv(archivo_salida, index=False)
    print(f"Archivo guardado exitosamente como: {archivo_salida}")
    
    # Opcional: devolver el DataFrame por si quieres seguir usándolo en Jupyter
    return df_filtrado
```

```python
# --- Ejemplo de cómo usar la función en tu Jupyter Notebook ---
archivo_origen = 'individual_proteins_dataset.csv'
archivo_destino = 'liposomes_dataset-np_type.csv'
atributo_categorico = 'np_type'
categoria = 'liposome'

# Llamar a la función
df_resultado = filtrar_liposomas(archivo_origen, archivo_destino, atributo_categorico, categoria)

# Mostrar las primeras filas del resultado en pantalla
df_resultado.head()
```
